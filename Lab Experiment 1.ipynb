```python
import pandas as pd
import matplotlib.pyplot as plt
data = {'Year': list(range(2000, 2011)),
'Job Postings': [100, 200, 380, 620, 830, 1250, 1650, 2050, 2600, 3400, 4300]}

df = pd.DataFrame(data)
plt.plot(df['Year'], df['Job Postings'], marker='o')
plt.title('Trend of Data Science Job Postings')
plt.xlabel('Year')
plt.ylabel('Number of Job Postings')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {'Role': ['Data Analyst', 'Data Scientist', 'Data Engineer',
             'Machine Learning Engineer', 'Business Analyst'],
    'Count': [120, 180, 100, 80, 70]}

df = pd.DataFrame(data)

plt.bar(df['Role'], df['Count'])
plt.title('Distribution of Data Science Roles')
plt.xlabel('Job Roles')
plt.ylabel('Number of Employees')
plt.xticks(rotation=20)
plt.show()
```

```python
import pandas as pd

structured_data = {
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [22, 25, 24]
}

unstructured_data = {
    'Text': [
        'Data Science is interesting.',
        'Python is easy to learn.',
        'Big Data is the future.'
    ]
}

semi_structured_data = {
    'Student': [
        {'ID': 1, 'Name': 'Alice', 'Marks': 90},
        {'ID': 2, 'Name': 'Bob', 'Marks': 85},
        {'ID': 3, 'Name': 'Charlie', 'Marks': 88}
    ]
}

df1 = pd.DataFrame(structured_data)
df2 = pd.DataFrame(unstructured_data)
df3 = pd.DataFrame(semi_structured_data)

print("Structured Data")
print(df1)

print("\nUn-Structured Data")
print(df2)

print("\nSemi-Structured Data")
print(df3)
```

```python
from cryptography.fernet import Fernet

key = Fernet.generate_key()

cipher = Fernet(key)

data = "My Password is 12345"

encrypted_data = cipher.encrypt(data.encode())

decrypted_data = cipher.decrypt(encrypted_data).decode()

print("Original Data:", data)
print("Encrypted Data:", encrypted_data)
print("Decrypted Data:", decrypted_data)
```
